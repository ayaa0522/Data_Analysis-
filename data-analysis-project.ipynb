```python
df['order_date'] = pd.to_datetime(df['order_date'])
print(df['order_date'])
```

```python
invalid_quantity = df[df['quantity'] < 1]
print(invalid_quantity)


invalid_price=(df[df['price'] <= 0])
print(invalid_price)

invalid_discount=(df[(df['discount'] < 0) | (df['discount'] > 1)])
print(invalid_discount)


invalid_shipping_cost=(df[df['shipping_cost'] < 0])
print(invalid_shipping_cost)


invaled_age=df[(df['customer_age'] < 18) | (df['customer_age'] > 69)]
print(invaled_age)



invalid_delivery_day=df[df['delivery_time_days'] > 10]
print(invalid_delivery_day)
```

```python
Q1 = df['profit_margin'].quantile(0.25)
Q3 = df['profit_margin'].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[ (df['profit_margin'] < lower_bound) |(df['profit_margin'] > upper_bound)]
print(outliers)
```
